```python
import spacy
```

```python
#  1 Tokenization
# Load the French language model
nlp = spacy.load("fr_core_news_sm")
texte = "Le professeur O. Mahmoudi enseigne le NLP à l'EST de Nador en 2026. Les étudiants aiment ce cours !"
doc = nlp(texte)
print("--- 1. Tokenization ---")
tokens = [token.text for token in doc]
print(tokens)
```

```python
#2. Segmentation des Phrases

nlp = spacy.load("fr_core_news_sm")

texte = "Le professeur O. Mahmoudi enseigne le NLP à l'EST de Nador en 2026. Les étudiants aiment ce cours !"
doc = nlp(texte)

print("--- 2. Segmentation des Phrases ---")
# Extract sentences
phrases = [phrase.text for phrase in doc.sents]
for i, phrase in enumerate(phrases):
    print("Phrase {",i+1,"}:" ,phrase)
```

```python
#3. Normalisation

print("--- 3. Normalisation ---")
texte = "Le professeur O. Mahmoudi enseigne le NLP à l'EST de Nador en 2026. Les étudiants aiment ce cours !"

# Convert to lowercase
texte_normalise = texte.lower()
print(texte_normalise)
```

```python
# 4. Reconnaissance d'entités nommées (NER)

nlp = spacy.load("fr_core_news_sm")

texte = "Le professeur O. Mahmoudi enseigne le NLP à l'EST de Nador en 2026. Les étudiants aiment ce cours !"
doc = nlp(texte)

print("--- 4. Reconnaissance d'entités nommées (NER) ---")
# Extract entities and their labels
entites = [(ent.text, ent.label_) for ent in doc.ents]
print(entites)
```

```python
#   5. Les mots vides (Stopwords)

nlp = spacy.load("fr_core_news_sm")

texte = "Le professeur O. Mahmoudi enseigne le NLP à l'EST de Nador en 2026. Les étudiants aiment ce cours !"
doc = nlp(texte)

print("--- 5. Les mots vides (Stopwords) ---")
# Filter out stopwords and punctuation
mots_importants = [token.text for token in doc if not token.is_stop and not token.is_punct]
print(mots_importants)
```

```python
# 6. Matchers en NLP 

from spacy.matcher import Matcher

nlp = spacy.load("fr_core_news_sm")
matcher = Matcher(nlp.vocab)

texte = "Le professeur O. Mahmoudi enseigne le NLP à l'EST de Nador en 2026. Les étudiants aiment ce cours !"
doc = nlp(texte)

print("--- 6. Matchers en NLP ---")
# Define a pattern for a 4-digit number (e.g., a year)
pattern = [{"SHAPE": "dddd"}]
matcher.add("ANNEE_PATTERN", [pattern])

matches = matcher(doc)
for match_id, start, end in matches:
    # context_id / dubi le mots / fi le mots/ in ... foction ---> (matcher) lktfrad 3lina nkadmo bihat 3 dyal les valeurs
    matched_span = doc[start:end]
    print(f"Motif trouvé : '{matched_span.text}' à la position {start}")
    
```

```python
# 7. La structure syntaxique (POS Tagging)

nlp = spacy.load("fr_core_news_sm")

texte = "Le professeur O. Mahmoudi enseigne le NLP à l'EST de Nador en 2026. Les étudiants aiment ce cours !"
doc = nlp(texte)

print("--- 7. La structure syntaxique (POS Tagging) ---")
# Extract tokens and their part-of-speech tags
pos_tags = [(token.text, token.pos_) for token in doc]

for mot, tag in pos_tags[:10]: # Print the first 10 for brevity
    print("Mot:" ,mot, '-'*15 , "> Tag POS:" ,tag)
```
