```python
from gensim.models import Word2Vec
import numpy as np
```

```python
corpus = [
    "Je mange une pomme rouge".lower().split(),
    "Je bois un jus de pomme".lower().split(),
    "Une pomme est un fruit".lower().split(),
    "La banane est un fruit jaune".lower().split(),
    "Je mange une banane".lower().split()
]
```

```python
corpus
```

```python
modele = Word2Vec(sentences=corpus, vector_size=10, window=2, min_count=1, sg=0)
```

```python
vecteurs_phrases = []
```

```python
print("--- Vecteurs des phrases ---")
for i, phrase in enumerate(corpus):
    # Récupération des vecteurs pour chaque mot de la phrase présent dans le modèle
    vecteurs_mots = [modele.wv[mot] for mot in phrase if mot in modele.wv]
    
    if vecteurs_mots:
        # Calcul de la moyenne sur l'axe 0 (المتوسط الحسابي)
        vecteur_phrase = np.mean(vecteurs_mots, axis=0)
    else:
        # Vecteur nul de secours si aucun mot n'est trouvé
        vecteur_phrase = np.zeros(modele.vector_size)
```

```python
vecteurs_phrases.append(vecteur_phrase) 
```

```python
print(f"\nPhrase {i+1} : {' '.join(phrase)}")
print(f"Vecteur (dimension {modele.vector_size}) : \n{vecteur_phrase}")
```
