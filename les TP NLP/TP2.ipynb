```python
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

rows, cols = img_rgb.shape[:2] 

# ----------------- 1.   (Translation) -----------------
 
M_translation = np.float32([[1, 0, 300], [0, 1, 255]])

img_translated = cv.warpAffine(img_rgb, M_translation, (cols, rows))

# ----------------- 2.   (Rotation) -----------------
center = (cols / 2, rows / 2)

M_rotation_45 = cv.getRotationMatrix2D(center, 45, 1)

img_rotated_45_linear = cv.warpAffine(img_rgb, M_rotation_30, (cols, rows), flags=cv.INTER_LINEAR)

M_rotation_90 = cv.getRotationMatrix2D(center, 90, 1)

img_rotated_90_cubic = cv.warpAffine(img_rgb, M_rotation_90, (cols, rows), flags=cv.INTER_CUBIC)
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 4, 1)
plt.imshow(img_rgb)
plt.title("Image Originale")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(img_translated)
plt.title("Translation (50px à droite)")
plt.axis("off")

plt.subplot(1, 4, 3)
plt.imshow(img_rotated_45_linear)
plt.title("Rotation 45° (Linéaire)")
plt.axis("off")

plt.subplot(1, 4, 4)
plt.imshow(img_rotated_90_cubic)
plt.title("Rotation 90° (Cubique)")
plt.axis("off")

plt.show()
```
