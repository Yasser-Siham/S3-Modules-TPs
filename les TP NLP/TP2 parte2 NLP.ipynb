```python
import spacy
nlp = spacy.load("en_core_web_sm") # Aw fr_core_news_sm l'francais
doc = nlp(texte_lower)

# N-khtarou ghir l-kalimat li machi ponctuation (!token.is_punct)
tokens_sans_punct = [token.text for token in doc if not token.is_punct]
print("\n2. Sans Ponctuation:")
print(" ".join(tokens_sans_punct))
```

```python
abbreviations = {
    "u.s.a.": "usa",
    "o.": "omayma" # Mital
}

#texte_sans_punct = " ".join(tokens_sans_punct)
for abbr, full in abbreviations.items():
    texte_sans_punct = texte_sans_punct.replace(abbr, full)

print("\n3. Sans Abréviations (Normalisé):")
print(texte_sans_punct)
```

```python
import re

# N-3awdo ay raqam (awla raqam fih fasila b7al 1,250) b <NUM>
texte_nombres = re.sub(r'\d+(,\d+)?', '<NUM>', texte_sans_punct)

print("\n4. Gestion des Nombres (Masquage):")
print(texte_nombres)
```

```python
import spacy
from collections import Counter

# 1. Chargement dyal l'modèle (Anglais) w tahlil n-nass
nlp = spacy.load("en_core_web_sm")
texte = "Apple Inc. was founded by Steve Jobs, Steve Wozniak, and Ronald Wayne in Cupertino, California, on April 1, 1976. The company launched the iPhone in 2007 and is now worth over $3 trillion. Tim Cook is the current CEO."
doc = nlp(texte)

# 2. Affichage dyal ga3 les entités
print("--- 2. Ga3 les entités ---")
for ent in doc.ents:
    print("L'entité :", ent.text, " | Type :", ent.label_)

# 3. L-7issab (Compte) dyal les types
print("\n--- 3. Ch7al mn entité f kulla type ---")
# Kanjm3ou ga3 les labels f wa7d l-liste
types_entites = [ent.label_ for ent in doc.ents]
compteur = Counter(types_entites)
for type_ent, nombre in compteur.items():
    print(type_ent, ":", nombre)

# 4. Affichage ghir dyal ORG w PERSON
print("\n--- 4. Ghir ORG w PERSON ---")
for ent in doc.ents:
    if ent.label_ in ["ORG", "PERSON"]:
        print(ent.text, "(", ent.label_, ")")
```

```python
import spacy

# 1. Chargement dyal l'modèle spaCy w tahlil n-nass
nlp = spacy.load("en_core_web_sm")
texte = "The quick brown fox jumps over the lazy dog. This sentence contains many common words that do not add much meaning."
doc = nlp(texte)

# 2. L-liste dyal ga3 les stopwords li kynin f l'anglais
stopwords_anglais = nlp.Defaults.stop_words
print("--- 2. Ch7al mn stopword kayn f spaCy (Anglais) ---")
print("L-majmou3 :", len(stopwords_anglais))

# 3. L-kalimat li machi stopwords (Mots utiles)
mots_utiles = [token.text for token in doc if not token.is_stop and token.is_alpha]
print("\n--- 3. L-kalimat l-mohima (Qbel ma nzidou stopwords jdad) ---")
print(mots_utiles)

# 4. Nzidou "sentence" w "add" l-stopwords dyal spaCy
nlp.Defaults.stop_words.add("sentence")
nlp.Defaults.stop_words.add("add")
# Khassna n3lmo l-vocabulaire bli had joj kalimat wellaw stopwords
nlp.vocab["sentence"].is_stop = True
nlp.vocab["add"].is_stop = True

# N3awdou n7ellou n-nass mn jdid
doc_jdid = nlp(texte)
mots_utiles_jdad = [token.text for token in doc_jdid if not token.is_stop and token.is_alpha]

print("\n--- 4. L-kalimat l-mohima (Be3d ma zidna 'sentence' w 'add') ---")
print(mots_utiles_jdad)
```

```python
import spacy

# 1. Chargement dyal l'modèle w tahlil n-nass
nlp = spacy.load("en_core_web_sm")
texte = "Apple launched the new iPhone."
doc = nlp(texte)

print("--- Tahlil dyal L-3alaqat N-Na7wiyya (Dependencies) ---")
print("L-Kalima", " | ", "L-3alaqa (dep_)", " | ", "L-Kalima L-Omm (head)")
print("-" * 50)

# 2. Ndb3ou l-i3rab w l-3alaqat l kulla kalima
for token in doc:
    print(token.text, " | ", token.dep_, " | ", token.head.text)

print("\n--- Chkon howa L-Jder (ROOT) dyal l-jomla? ---")
for token in doc:
    if token.dep_ == "ROOT":
        print("L-fi3l l-asassi howa :", token.text)
```

```python
import spacy
from nltk.stem import PorterStemmer

# 1. N-wjdou l-Adawat
nlp = spacy.load("en_core_web_sm")
ps = PorterStemmer()

# L-kalimat li baghin n-tstiw
mots = ["running", "played", "easily", "mice", "better"]
# Jm3nahom f jomla bach spaCy yqrahom
texte = " ".join(mots)
doc = nlp(texte)

print("--- Moqaranat Stemming vs Lemmatisation ---")
print("L-kalima", " | ", "Stemming (Porter)", " | ", "Lemmatisation (spaCy)")
print("-" * 60)

# 2. N-tb9ou l-Amaliyat b-jouj 3la kulla kalima
for token in doc:
    # A. Stemming (Kaydiro NLTK)
    stem_result = ps.stem(token.text)
    
    # B. Lemmatisation (Kaydirha spaCy, w hiya wajda f token.lemma_)
    lemma_result = token.lemma_
    
    # C. N-tb3ou n-Natija
    print(token.text, " | ", stem_result, " | ", lemma_result)
```
