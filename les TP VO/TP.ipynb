```python
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np
#img_bgr = cv.imread('image_test3.png')
```

```python
img = cv.imread('image_test4.jpeg')
img.shape
```

```python
print('min :',np.min(img), 'max :', np.max(img))
print(np.where(img == np.max(img)))
#print( cv.THRESH_BINARY(src='img' , thresh=127 , maxval = 255))
```

```python
plt.hist(img.ravel() , 255 ,[0 , 256])
plt.xlim([0,256])
```

```python
ret , img_bin = cv.threshold(img, 127, 255, cv.THRESH_BINARY)
```

```python
ret 
```

```python
#img_bin
```

```python
#apres souilage
plt.subplot(1,2,1)
plt.imshow(img_bin,cmap='gray')
# avent souilage
plt.subplot(1,2,2)
plt.imshow(img,cmap='gray')
```

```python
min_var = np.min(img)
max_var = np.max(img)
img_float = img.astype(float) 
img_norm = ((img_float - min_var)/ max_var - min_var)*255
img_norm
```

```python
img_norm = img.astype(np.uint8)
img_norm
```

```python
#apres souilage
plt.subplot(1,2,1)
plt.imshow(img,cmap='gray')
plt.title('avent normalisation')
# avent souilage
plt.subplot(1,2,2)
plt.imshow(img_norm, cmap='gray')
plt.title('apres normalisation')
```

```python
# utilise Égalisation acec finction cv.equalizeHist
img_eq = cv.equalizeHist(img)
```

```python
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Image originale")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(img_eq, cmap='gray')
plt.title("Image avec Égalisation")
plt.axis("off")
```
