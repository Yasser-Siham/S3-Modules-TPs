```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python

img = cv2.imread("img.png")

# 1. Grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
# 2. Gaussian Blur
blur = cv2.GaussianBlur(gray, (5, 5), 0)

# 3. Gradient Sobel
gx = cv2.Sobel(blur, cv2.CV_64F, 1, 0, ksize=3)
gy = cv2.Sobel(blur, cv2.CV_64F, 0, 1, ksize=3)
gradient = cv2.magnitude(gx, gy)
```

```python

# 4. Canny (Non-Maximum Suppression,
# Double Threshold, Hysteresis)
edges = cv2.Canny(blur, 50, 150)

# Affichage
images = [gray, blur, gradient, edges]
titles = ["Grayscale", "Gaussian", "Sobel", "Canny"]
```

```python







plt.figure(figsize=(12, 4))
for i in range(4):
    plt.subplot(1, 4, i + 1)
    plt.imshow(images[i], cmap="gray")
    plt.title(titles[i])
    plt.axis("off")
plt.tight_layout()
plt.show()
```

Watershed

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Lire l'image en couleur et en gris
img = cv2.imread("img.png")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 1. Seuillage Otsu
_, binary = cv2.threshold(
    gray, 0, 255,
    cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU
)

# 2. Transformation de distance
dist = cv2.distanceTransform(binary, cv2.DIST_L2, 5)

# 3. Détection des marqueurs
_, fg = cv2.threshold(dist, 0.5 * dist.max(), 255, 0)
fg = np.uint8(fg)

bg = cv2.dilate(binary, np.ones((3, 3), np.uint8), iterations=3)
unknown = cv2.subtract(bg, fg)

_, markers = cv2.connectedComponents(fg)
markers = markers + 1
markers[unknown == 255] = 0

# 4. Watershed
markers = cv2.watershed(img.copy(), markers)

# 5. Colorier les contours en rouge
result = img.copy()
result[markers == -1] = (0, 0, 255)

# Affichage
images = [binary, dist, markers,
          cv2.cvtColor(result, cv2.COLOR_BGR2RGB)]

titles = ["Seuillage", "Distance Transform",
          "Watershed", "Résultat final"]

plt.figure(figsize=(12, 4))

for i in range(4):
    plt.subplot(1, 4, i + 1)
    plt.imshow(images[i], cmap=None if i == 3 else "gray")
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```
